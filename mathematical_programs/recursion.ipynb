### What is Recursion, Really? (रिकर्सन असल में क्या है?)
* English: Recursion is a technique where a function calls itself to solve a smaller piece of the exact same problem, until it hits a known simple condition (called the Base Case).
* Hinglish: रिकर्सन का मतलब है—जब एक फंक्शन खुद को ही दोबारा कॉल करे। लेकिन बड़ी समस्या को एक ही झटके में हल करने के बजाय, यह उसे छोटे टुकड़ों में तोड़ता जाता है और आखिर में एक ऐसी छोटी स्थिति पर रुक जाता है जिसका जवाब सीधे पता होता है।

#### The Golden Rule (दो जरूरी चीजें जो हर रिकर्सन में होनी चाहिए):
* Base Case (रुकने की शर्त): अगर बेस केस नहीं होगा, तो फंक्शन अनंत बार (Infinite loop) खुद को कॉल करेगा और कंप्यूटर की रैम फुल हो जाएगी (Stack Overflow एरर आ जाएगा)।
* Recursive Step (छोटा करने का कदम): हर बार फंक्शन को खुद को ऐसे कॉल करना चाहिए कि प्रॉब्लम का साइज़ छोटा होता जाए और वह बेस केस की तरफ बढ़े।

#### Example 1: Factorial of a Number (n!)
* English: The factorial of a number n is the product of all positive integers less than or equal to n (e.g., 5! = 5 x 4 x 3 x 2 x 1 = 120). Mathematically, we can write it as: n! = n x (n-1)!.
* Hinglish: किसी नंबर का फैक्टोरियल मतलब उस संख्या से लेकर 1 तक के सभी नंबरों का गुणा। गणित के हिसाब से: 5! का मतलब है $5 x (4!)। यानी हमने बड़े सवाल (5!) को छोटे सवाल (4!) में बदल दिया!

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n - 1)

factorial(5)
```

#### 🧠 Tracing & Call Stack Visualization (फैक्टोरियल का ड्राई-रन):
जब आप factorial(5) चलाएंगे, तो कंप्यूटर की मेमोरी (Call Stack) में यह प्रक्रिया इस तरह होगी:
1. factorial(5) कॉल हुआ -> यह रुकता है और इंतज़ार करता है factorial(4) के जवाब का। (5 x wait)
2. factorial(4) कॉल हुआ -> यह इंतज़ार करता है factorial(3) के जवाब का। (4 x wait)
3. factorial(3) कॉल हुआ -> यह इंतज़ार करता है factorial(2) के जवाब का। (3 x wait)
4. factorial(2) कॉल हुआ -> यह इंतज़ार करता है factorial(1) के जवाब का। (2 x wait)
5. factorial(1) कॉल हुआ -> यहाँ Base Case मिल गया! यह सीधे 1 रिटर्न कर देता है।
##### अब स्टैक उल्टा खुलेगा (Unwinding phase):
* factorial(1) ने दिया -> 1
* factorial(2) के पास गया 2 x 1 = 2
* factorial(3) के पास गया 3 x 2 = 6
* factorial(4) के पास गया 4 x 6 = 24
* सबसे ऊपर factorial(5) के पास आया $5 x 24 = 120 (फाइनल जवाब!)

### Example 2: Fibonacci Series (फिबोनाची सीरीज)
* English: The Fibonacci sequence is a series of numbers where each number is the sum of the two preceding ones. Starting from 0 and 1, the sequence goes: 0, 1, 1, 2, 3, 5, 8, 13, ...Mathematically: F(n) = F(n-1) + F(n-2)
* Hinglish: फिबोनाची सीरीज में हर अगला नंबर अपने पिछले दो नंबरों को जोड़कर बनता है।जैसे: 0, 1, (0+1)=1, (1+1)=2, (1+2)=3, (2+3)=5 ...यानी F(4) = F(3) + F(2)।
