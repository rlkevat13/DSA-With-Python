### What is LCM? (एलसीएम क्या है?)
* English: LCM stands for Least Common Multiple. It is the smallest positive integer that is perfectly divisible by two or more given numbers (without leaving any remainder).

* Hinglish: LCM का मतलब है—वह सबसे छोटा कॉमन नंबर जो दी गई दोनों संख्याओं से पूरी तरह भाग हो जाए (यानी दोनों के पहाड़े में सबसे पहले कॉमन आए)।

### Real-World Example (असली जिंदगी का उदाहरण):
* English: Imagine two runners on a circular track. Runner A completes one round in 4 minutes, and Runner B completes one round in 6 minutes. If they start running together at the exact same time, after how many minutes will they meet again at the starting line?
* Hinglish: मान लीजिए दो धावक एक गोल ट्रैक पर दौड़ रहे हैं। पहला धावक एक चक्कर 4 मिनट में पूरा करता है और दूसरा 6 मिनट में। अगर दोनों एक साथ दौड़ना शुरू करें, तो वे पहली बार दोबारा स्टार्टिंग लाइन पर कितने मिनट बाद मिलेंगे?
* Calculation (कैसे निकालेंगे?): 
    * 4 के गुणज (Multiples / पहाड़ा): 4, 8, 12, 16, 20, 24, ...
    * 6 के गुणज (Multiples / पहाड़ा): 6, 12, 18, 24, 30, ...
    * दोनों में कॉमन मल्टीपल्स: $12, 24, 36, ...
    * इनमें सबसे छोटा (Least) कौन सा है? 12।
    * इसका मतलब, वे दोनों ठीक 12 मिनट बाद पहली बार दोबारा एक साथ मिलेंगे! इसलिए 4 और 6 का LCM = 12 है।

```python
def compute_lcm(x, y):
    if x > y:
        greater = x
    else:
        greater = y
    while True:
        if (greater % x == 0) and (greater % y == 0):
            lcm = greater
            break
        greater += 1
    return lcm

compute_lcm(54, 24)
```

```python
import math

result = math.lcm(4, 6)
print(f"4 aur 6 ka LCM hai: {result}")
```

#### Using the GCD Formula (लॉजिक समझने के लिए) (Best way)

```python
def find_gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

def find_lcm(a, b):
    if a == 0 or b == 0:
        return 0
    return abs(a * b) // find_gcd(a, b)

find_lcm(12, 18)
```
