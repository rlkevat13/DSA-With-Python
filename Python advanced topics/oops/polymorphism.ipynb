### Polymorphism (Ek Roop, Anek Kaam)
* English: Polymorphism means "many forms". It allows different classes to have methods with the same name, but they can behave differently.

* Hinglish: Iska matlab hai "ek hi cheez ke alag-alag roop". Jaise ek hi insaan office mein employee hai, ghar par beta hai, aur market mein customer hai. Code mein, agar alag-alag classes mein same naam ka function ho, toh object ke hisab se apna kaam karta hai.

```python
class Cat:
    def make_sound(self):
        print("Meow Meow")

class Duck:
    def make_sound(self):
        print("Quack Quark")

def animal_sound(animal_object):
    animal_object.make_sound()

c = Cat()
d = Duck()

animal_sound(c)
animal_sound(d)
```

#### Polymorphism & Abstraction (Advanced Concepts)
* Method Overriding: Jab child class apne parent class ke method ko apne tarike se rewrite kar deti hai.

* Operator Overloading (Dunder/Magic Methods): Python mein +, -, * operators ko apni classes ke liye customize karna (jaise __add__, __str__).

* Abstraction (abc module): Blueprint enforce karna. Agar koi class abstract hai, toh uske methods ko child class mein implement karna mandatory hota hai.

```python
class Vector:
    def __init__(self, x):
        self.x = x

    # Operator Overloading for '+'
    def __add__(self, other):
        return Vector(self.x + other.x)

    def __str__(self):
        return f"Vector({self.x})"

v1 = Vector(10)
v2 = Vector(20)
print(v1 + v2)
```

### Polymorphism & Abstraction (Deep Sub-topics)
#### Sub-topics to prepare:
* Method Overriding: Child class ka parent ke method ko overwrite karna.

* Duck Typing: "If it walks like a duck and quacks like a duck, it's a duck." Python mein type check karne ki zaroorat nahi hoti, bas method hona chahiye.

* Operator Overloading (Magic/Dunder Methods): +, -, == jaise operators ko custom classes par chalana (__add__, __eq__).

* Abstract Base Classes (ABC): Incomplete classes banana jinke methods child class ko implement karne hi padte hain.

#### Interview Explanation:
* English:

    * "Polymorphism allows different types to be treated via the same interface. In Python, this is heavily supported by Duck Typing, where object behavior matters more than its class inheritance. Abstraction hides implementation using the abc module, forcing child classes to implement specific abstract methods."

* Hinglish:

    * Duck Typing: Python mein is baat se fark nahi padta ki object kis class ka hai, fark isse padta hai ki uske paas required method hai ya nahi.

    * Operator Overloading: Aap apne custom objects par + ya * operator chala sakte hain dunder methods (__add__) use karke.

```python
# 1. Duck Typing Example
class PDFDocument:
    def price_doc(self):
        print("Printing PDF file...")

class WordDocument:
    def price_doc(self):
        print("Printing Word file...")

# It doesn't care about class type, it only cares if 'print_doc' exists
def handle_printing(doc_object):
    doc_object.print_doc()

handle_printing(PDFDocument())
handle_printing(WordDocument())
```

```python
# 2. Operator Overloading Example (Magic Methods)
class Cart:
    def __init__(self, total_amount):
        self.total_amount = total_amount

    # Overloading the '+' operator using __add__
    def __add__(self, other):
        return Cart(self.total_amount + other.total_amount)

    def __str__(self):
        return f"Total Cart Amount: {self.total_amount}"

c1 = Cart('500')
c2 = Cart('1200')
print(c1 + c2)
```

#### Polymorphism (Advanced Sub-topics)
Polymorphism ke andar yeh 3 main sub-topics hote hain jo interviewers aksar poochte hain:
##### A. Method Overriding
* Interview Definition (English):

    "Method Overriding occurs when a child class provides a specific implementation of a method that is already defined in its parent class. At runtime, the child's method takes precedence and overrides the parent's version."

* Hinglish mein samjhein:
    Jab parent class ke paas ek method ho, aur child class same naam ka method apne tarike se likh kar usko overwrite kar de, toh use Method Overriding kehte hain.

```python
class Animal:
    def make_sound(self):
        print("Some generic animal sound...")

class Dog(Animal):
    # Overriding the parent's make_sound method
    def make_sound(self):
        print("Bark! Bark!")

d = Dog()
d.make_sound()
```

##### B. Duck Typing
* Interview Definition (English):

    "Duck Typing is a concept in dynamic languages like Python where the type or class of an object is less important than the methods it defines. As the saying goes: 'If it walks like a duck and quacks like a duck, it's a duck.' Python does not check the explicit type; it only checks if the required method exists."

* Hinglish mein samjhein:
    Python mein is baat se koi matlab nahi hota ki object kis class ka hai. Agar uske paas zaroori function (method) maujood hai, toh Python usko run kar dega. Type checking ki zaroorat nahi padti.

```python
class PDFExporter:
    def export(self):
        print("Exporting data to PDF format...")

class ExcelExporter:
    def export(self):
        print("Exporting data to Excel format...")

# Duck Typing in action: It doesn't care about the class name, only the .export() method
def perform_export(exporter_obj):
    exporter_obj.export()

perform_export(PDFExporter())
perform_export(ExcelExporter())
```

##### C. Operator Overloading (Magic / Dunder Methods)
* Interview Definition (English):

    "Operator Overloading allows developers to redefine how built-in operators (like +, -, ==) behave with user-defined objects by implementing special magic or dunder methods (e.g., __add__, __eq__)."

* Hinglish mein samjhein:
    Aap Python ke standard operators (+, -, *) ko apni custom classes ke objects par bhi chala sakte hain, bas aapko uske liye dunder methods (__add__) likhne padte hain.

```python
class ShoppingCart:
    def __init__(self, amount):
        self.amount = amount

    # Overloading the '+' operator using __add__ megic method
    def __add__(self, other):
        return ShoppingCart(self.amount + other.amount)

    def __str__(self):
        return f"Total Amount: {self.amount}"

cart1 = ShoppingCart(250)
cart2 = ShoppingCart(450)

# Yeh internaly cart1.__add__(cart2) call karega
print(cart1 + cart2)
```

##### Abstraction (Complete Concept & Example)
* Interview Definition (English):

    "Abstraction is the process of hiding internal implementation details and exposing only the essential interface to the user. In Python, abstraction is implemented using the built-in abc (Abstract Base Classes) module. An abstract class cannot be instantiated directly and can contain abstract methods that must be implemented by all its child classes, enforcing a strict structural blueprint."

* Hinglish mein samjhein:
    Abstraction ka matlab hai complex code aur inner logic ko chhupana aur baahar wale ko sirf zaroori buttons ya methods dena. Python mein abc module ka use karke hum aisi class banate hain jiska object direct nahi ban sakta, aur uske andar ke abstract methods ko child class mein override karna compulsory hota hai.

```python
from abc import ABC, abstractmethod

# Abstract Base Class (Blueprint)
class PaymentGateway(ABC):
    @abstractmethod
    def authenticate(self):
        pass

    @abstractmethod
    def make_payment(self, amount):
        pass

# Child class implementing the blueprint
class PhonePe(PaymentGateway):
    def authenticate(self):
        print("PhonePe UPI PIN verified successfully!")

    def make_payment(self, amount):
        print(f"Paid ₹{amount} using PhonePe.")

# --- Usage ---
# gateway = PaymentGateway() # Error! You cannot instantiate an abstract class directly.
payment = PhonePe()
payment.authenticate()
payment.make_payment(1200)
```

##### Method Overloading (aur Python ka Behavior)
* Interview Definition (English):

    "Method Overloading is the ability to define multiple methods with the same name but different parameters within the same class. Python does not support traditional method overloading natively; if you define multiple methods with the same name, the last defined method overrides the previous ones. However, we achieve similar behavior in Python using default arguments, variable-length arguments (*args), or functools.singledispatch."

* Hinglish mein samjhein:
    C++ ya Java mein ek hi class ke andar same naam ke multiple functions bana sakte hain bas unke parameters alag hote hain. Par Python strict method overloading support nahi karta—agar aap do same naam ke function bana doge, toh aakhri wala pehle wale ko overwrite kar dega. Python mein hum iska kaam default arguments ya *args use karke chalate hain.

```python
class Calculator:
    # Python mein overloading ke liye hum default arguments ya None use karte hain
    def add(self, a, b, c=None):
        if c is not None:
            return a + b + c
        return a + b

calc = Calculator()
print(calc.add(10, 20))
print(calc.add(10, 20, 30))
```
