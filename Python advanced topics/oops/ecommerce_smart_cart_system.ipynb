```python
from abc import ABC, abstractmethod


# ==========================================
# 1. ABSTRACTION & POLYMORPHISM (Payment Gateway)
# ==========================================
class PaymentGateway(ABC):
  """Abstract base class enforcing a strict blueprint for all payment methods."""

  @abstractmethod
  def authenticate(self):
    pass

  @abstractmethod
  def process_payment(self, amount):
    pass


class UPIPayment(PaymentGateway):

  def authenticate(self):
    print("UPI PIN verified securely!")

  def process_payment(self, amount):
    print(f"Paid ₹{amount} successfully via UPI.")


# ==========================================
# 2. ENCAPSULATION (Product with Private Price)
# ==========================================
class Product:

  def __init__(self, name, price, category):
    self.name = name
    self.__price = price  # Private variable triggering Name Mangling
    self.category = category

  @property
  def price(self):  # Getter method using @property[cite: 3]
    return self.__price

  @price.setter
  def price(self, new_price):  # Setter with validation logic[cite: 3]
    if new_price > 0:
      self.__price = new_price
    else:
      print("Error: Price cannot be negative!")


# ==========================================
# 3. INHERITANCE & MIXIN (Discount & Logging Features)
# ==========================================
class LoggerMixin:
  """Mixin class providing independent logging behavior[cite: 2]."""

  def log_activity(self, action):
    print(f"[LOG - {self.__class__.__name__}]: {action}")


class DiscountedProduct(Product, LoggerMixin):
  """Inherits from Product and mixes in LoggerMixin."""

  def __init__(self, name, price, category, discount_percent):
    super().__init__(name, price, category)
    self.discount_percent = discount_percent

  def get_discounted_price(self):
    final_price = self.price - (self.price * self.discount_percent / 100)
    self.log_activity(f"Calculated discount for {self.name}")
    return final_price


# ==========================================
# 4. CLASSES, OBJECTS & OPERATOR OVERLOADING
# ==========================================
class ShoppingCart:
  shop_name = "PythonicMart"  # Class Variable

  def __init__(self, customer_name):
    self.customer_name = customer_name  # Instance Variable[cite: 1]
    self.items = []  # List to store (product, quantity)

  def add_item(self, product, quantity):
    self.items.append((product, quantity))

  def calculate_total(self):
    total = 0
    for product, qty in self.items:
      # Checking if product has discount method (Duck Typing behavior)
      if hasattr(product, "get_discounted_price"):
        total += product.get_discounted_price() * qty
      else:
        total += product.price * qty
    return total

  # Operator Overloading for '+' to merge two carts[cite: 4]
  def __add__(self, other_cart):
    merged_cart = ShoppingCart(
        f"{self.customer_name} & {other_cart.customer_name}"
    )
    merged_cart.items = self.items + other_cart.items
    return merged_cart

  def __str__(self):
    return (
        f"Cart Owner: {self.customer_name} | Total Items:"
        f" {len(self.items)}"
    )


# ==========================================
# -- EXECUTION / USAGE DEMO --
# ==========================================

# Creating products using Encapsulation & Inheritance[cite: 2, 3]
p1 = Product("Wireless Mouse", 1200, "Electronics")
p2 = DiscountedProduct("Mechanical Keyboard", 4500, "Electronics", 10)

# Creating Cart objects[cite: 1]
cart1 = ShoppingCart("Rahul")
cart1.add_item(p1, 2)

cart2 = ShoppingCart("Amit")
cart2.add_item(p2, 1)

# Merging carts using Operator Overloading (__add__)[cite: 4]
combined_cart = cart1 + cart2
print(combined_cart)
print(f"Grand Total Amount: ₹{combined_cart.calculate_total()}")

# Processing Payment via Abstraction (UPIPayment implements PaymentGateway)[cite: 4]
payment_gateway = UPIPayment()
payment_gateway.authenticate()
payment_gateway.process_payment(combined_cart.calculate_total())
```

### Is Project mein kaun-sa OOPs concept kahan use hua? (Quick Revision):
1. Classes, Objects & Instance/Class Variables (shop_name, customer_name): ShoppingCart aur Product classes ke zariye objects banaye gaye hain[cite: 1].

2. Encapsulation (__price, @property, @price.setter): Product ki price ko secure rakha gaya hai taaki koi direct negative price set na kar sake[cite: 3].

3. Inheritance & Mixins (DiscountedProduct, LoggerMixin, super()): Code reusability ke liye parent-child hierarchy aur independent logging mixin ka use kiya gaya hai[cite: 2, 3].

4. Operator Overloading (__ add __): Do alag-alag ShoppingCart objects ko aapas mein + operator se jod diya gaya hai[cite: 4].

5. Abstraction & Polymorphism (ABC, @abstractmethod, UPIPayment): Payment gateway ka ek strict blueprint banaya gaya hai jise alag-alag payment methods implement karte hain[cite: 4].
