### **Encapsulation:** Encapsulation is the practice of wrapping data (variables) and methods together inside a class and restricting direct access to safeguard internal data. In Python, we achieve this using private attributes (__variable) through Name Mangling, and manage them using getters and setters (@property).

### Hinglish mein samjhein:
Data ko safe rakhna. Jaise bank account ka balance koi direct change na kar sake, iske liye hum usko private (__balance) bana dete hain aur access ke liye methods ya @property decorator use karte hain.

### Encapsulation (Data Protection / Security)
* English: Encapsulation is the practice of hiding internal data and restricting direct access. We wrap data (variables) and methods together inside a class and protect sensitive data using private variables (using __).

* Hinglish: Jaise bank account mein aapka balance direct koi bahar wala nahi dekh sakta, uske liye ATM PIN ya password chahiye hota hai. Waise hi code mein sensitive data ko private (__) karke secure rakhna Encapsulation kehlata hai.

```python
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # Private Variable (hidden from outside)
    
    # Getter method to sefely access private data
    def get_balance(self):
        return self.__balance

acc = BankAccount("Rahul", 50000)
# print(acc.__balance)  # Error dega kyunki ye private hai!
print(acc.get_balance())
```

### Encapsulation & Property Decorators (Data Security)
* Access Modifiers in Python:

    * Public (var): Kahin se bhi access ho sakta hai.

    * Protected (_var): Convention ke taur par, sirf class aur uske subclasses ke liye (single underscore).

    * Private (__var): Name mangling hoti hai (_ClassName__var), class ke bahar se direct access nahi hota.

* Getters and Setters (@property): Private data ko safely read ya modify karne ke liye Pythonic tarika.

```python
class Product:
    def __init__(self, price):
        self.__price = price  # Private

    @property
    def price(self):  # Getter
        return self.__price

    @price.setter
    def price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price invalid hai!")

p = Product(100)
p.price = 15  # Setter call hua
print(p.price)  # Getter call hua -> 150
print(p._Product__price)  # Asli private variable check karte hain:
```

### Encapsulation (Deep Sub-topics)
Sub-topics to prepare:
* Access Specifiers in Python: Public (var), Protected (_var), Private (__var).

* Name Mangling: Python internally private variables ke naam ko kaise badal deta hai.

* Property Decorators (@property, @setter): Clean getters/setters banane ka Pythonic tarika.

### Interview Explanation:
* English:

    "Python doesn't have strict access modifiers like Java/C++. Instead, it uses naming conventions: single underscore for protected, and double underscore for private. Private variables undergo Name Mangling, where Python internally rewrites the variable name to _ClassName__variable to prevent accidental access from outside."

* Hinglish:
    Python mein strict private restrictions nahi hoti. Double underscore (__) lagane par Python Name Mangling karta hai aur uska naam badal kar _ClassName__variable kar deta hai taaki koi galti se bahar se access na kar le. Isko safely access karne ke liye hum @property decorators use karte hain.

```python
class EmployeeAccount:
    def __init__(self, salary):
        self.__salary = salary  # Private variable (Triggers Name Mangling)

    @property
    def salary(self):
        return self.__salary

    @salary.setter
    def salary(self, new_salary):
        if new_salary > 0:
            self.__salary = new_salary
        else:
            raise ValueError("Salary negative nahi ho sakti!")

emp = EmployeeAccount(50000)
print(emp.salary)  # Calls @property -> 50000

emp.salary = 60000  # Calls @setter
print(emp.salary)

print(
    emp._EmployeeAccount__salary
)
```

#### Deleter (@property_name.deleter)
* English Definition: Just like we use @property for getting and @setter for setting values, Python provides a .deleter decorator. It allows us to control what happens when someone tries to delete a private attribute using the del keyword.

* Hinglish mein samjhein: Jaise aap value dekhne ke liye getter aur badalne ke liye setter use karte hain, waise hi agar koi user del obj.price likh kar us variable ko delete karne ki koshish kare, toh aap us par control rakhne ke liye @price.deleter use karte hain.

```python
class Product:
    def __init__(self, price):
        self.__price = price

    @property
    def price(self):
        return self.__price

    @price.setter
    def price(self, new_price):
        if new_price > 0:
            self.__price = new_price
        else:
            print("Price invalid hai!")

    @price.deleter
    def price(self):
        print("Deleting price attribute...")
        del self.__price

p = Product(100)
del p.price  # Yeh deleter ko trigger karega
```
