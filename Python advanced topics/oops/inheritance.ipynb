#### **Inheritance:** Inheritance is a mechanism where a new class (child/ derived class) adopts the properties and methods of an existing class (parent/ base class), promoting code reusability.
* Type of inheritance:
    * Single Inheritance
    * Multilevel Inheritance
    * Hierarchical Inheritance
    * Multiple Inheritance
    * Hybrid Inheritance
#### **MRO:** MRO (Method resolution order) is the order in which Python looks for a method is a class hierarchy, determined by the C3 linearization algorithm.

#### **Single Inheitance:** In Single inheritance, a child class inherits properties and methods from a single parent class.

```python
class Animal:
    def eat(self):
        print("Yeh animal khana kha raha hai...")

class Dog(Animal):
    def bark(self):
        print("Dog bhauk raha hai: Bhau Bhau!")

d = Dog()
d.eat()
d.bark()
print(Dog.__mro__)
```

#### **Multilevel Inheritance:** In Multilevel inheritance, a feature is derived from a chain of inheritance. A child class becomes a parent for another class (Grandparent -> Parent -> Child).

```python
class Grandfather:
    def property_gf(self):
        print("Dadaji ki purani property.")

class Father(Grandfather):
    def property_father(self):
        print("papa ka apna ghar.")

# Child class inherite Father
class Son(Father):
    def property_son(self):
        print("Beta ki apni car.")

print(Son.__mro__)
s = Son()
s.property_gf()
s.property_father()
s.property_son()
```

#### **Hierarchical Inheritance:** In Hierarchical inheritance, multiple child classes inherit from a single parent class.

```python
class Vehical:
    def general_info(self):
        print("Yeh ek general vehical hai.")

class Car(Vehical):
    def drive_car(self):
        print("Car chal rahi hai road par.")

class Bike(Vehical):
    def ride_bike(self):
        print("Bike chal rahi hai road par.")

print(Car.mro())
c = Car()
c.general_info()
c.drive_car()

print("\n")

print(Bike.__mro__)
b = Bike()
b.general_info()
b.ride_bike()
```

#### **Multiple Inheritance:** In Multiple inheritance, a single child class inherits from more than one parent class. Python supports this seamlessly using the MRO (Method Resolution Order) algorithm.

```python
class Father:
    def skills_father(self):
        print("Father ki skill: Business management")

class Mother:
    def skills_mother(self):
        print("Mother ki skill: Creative painting")

class Child(Father, Mother):
    def skills_child(self):
        print("Child ki apni skill: Coding & AI")

print(Child.__mro__)
ch = Child()
ch.skills_father()
ch.skills_mother()
ch.skills_child()
```

#### **Hybrid Inheritance:** Hybrid inheritance is a combination of two or more type of inheritance (e.g., combining Hierarchical and Multiple Inheritance, Which often leads to the Diamand Problem).

```python
class A:
    def method_a(self):
        print("Method from Class A.")

class B(A):
    def method_b(self):
        print("Method from Class B.")

class C(A):
    def method_c(self):
        print("Method from Class C.")

class D(B, C):
    def method_d(self):
        print("Method from Class D.")

print(D.__mro__)
d = D()
d.method_a()
d.method_b()
d.method_c()
d.method_d()
```

#### **MRO:** Method Resolution Order (MRO) is the linear order in which Python searches for a method or attribute in a class hierarchy. Python computes this order using the C3 Linearization Algorithm (also known as Parker's Merge Algorithm). It ensures three critical properties: consistency with local precedence order, preservation of monotonicity, and handling of complex multiple inheritance graphs without ambiguity.

#### How C3 Linearization Decides the Next Class (The Rules):
Jab aap ek nayi class banate hain jo multiple classes ko inherit karti hai (jaise class D(B, C):), toh Python background mein ek formula run karta hai jisse hum C3 rules kehte hain.

Yeh rules teen main baaton par kaam karte hain:

1. Local Precedence Order (Child comes before Parents):

Agar D class B aur C se inherit kar rahi hai, toh MRO mein hamesha D sabse pehle aayega, phir uske parents (B aur C).

2. First-Come-First-Served (Parent Order):

Class definition ke dauran aapne jo order likha hai (class D(B, C)), Python usme left-to-right priority deta hai. Matlab B ko C se pehle priority milegi.

3. Monotonicity (No Breaking Parent Rules):

Agar kisi parent class ka apna koi MRO pehle se defined hai, toh child class us order ko kabhi tod nahi sakti. (Jaise agar B ke andar koi base class pehle aati hai, toh woh order maintain rahega).

#### Python iska MRO kaise calculate karta hai? (Step-by-Step Hinglish):
1. Sabse pehle current class: D khud sabse aage aayegi -> [D]

2. Phir Left parent check hoga (B): B ka MRO kya hai? [B, A, object]. Toh D ke baad B aayega -> [D, B]

3. Phir Right parent check hoga (C): C ka MRO kya hai? [C, A, object].

4. Merge operation (C3 Algorithm):

    * Python dekhega ki kya C ko B ke pehle rakha ja sakta hai? Nahi, kyunki B ki definition mein left priority thi.

    * Kya A ko direct la sakte hain? Nahi, kyunki A dono B aur C ka parent hai, isliye A tab tak nahi aa sakta jab tak uske bachhe (B aur C) process na ho jayein.

    * Isliye order banega: D -> B -> C -> A -> object.

#### What is the "Diamond Problem" and how MRO solves it?
#### The Diamond Problem is an ambiguity that arises in multiple inheritance when a class (D) inherits from two classes (B and C) that have a common base class (A). This forms a diamond-shaped inheritance graph. Without a strict ordering algorithm, the language wouldn't know whether to execute the base class (A) methods through B or through C, leading to duplicate execution or ambiguity.

#### The Diamond Problem occurs in multiple inheritance when a class inherits from two classes that share a common base class (forming a diamond shape). Without a deterministic algorithm like C3 Linearization, the program wouldn't know whether to inherit the base class method from the left parent or the right parent.

#### How MRO Solves the Diamond Problem
Purani languages (jaise C++) mein yeh ek badi problem thi jisme C++ ko virtual inheritance use karni padti thi. Par Python mein C3 Linearization algorithm isko automatically fix kar deta hai.

Python kya karta hai?

* Yeh saari classes ko ek linear sequence (ek hi line mein order) mein khada kar deta hai.

* Diamond structure (A top par, B & C beech mein, aur D niche) ko tod kar ek simple order bana deta hai: D -> B -> C -> A -> object.

* Isse koi bhi class do baar call nahi hoti, aur har method apne sahi sequence mein execute hota hai.
