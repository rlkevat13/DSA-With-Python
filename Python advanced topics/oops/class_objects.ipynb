#### **Class:** A Class is a user-defined blueprint or prototype from which objects are created. It defines a set of attributes (variables) and methods (functions).
#### **Object:** A object is an instance of a class that represents a real world entity with state and behavior.
#### **Self:** Self is a reference to the current instance of the class, used to access variables belonging to the class.
#### **Constractor:** A constractor (__ init __) is a special magic method that is automatically invoked when an object of a class is instantiated. Its primary role is to initialize the instance variables of that object.

```python
# Class definition
class Car:
    def __init__(self, brand, color):
        self.brand = brand  # Instance variable
        self.color = color

    def start_engine(self):
        print(f"{self.brand} ki gadi start ho gayi! Vrooom...")

# Creating objects (Instance)
car1 = Car("Hyundai", "White")
car2 = Car("Honda", "Black")

car1.start_engine()  # Output: Hyundai ki gadi start ho gayi! Vrooom...
print(car2.color)  #Output: Black
```

#### **Destructor:** A destructor (__ del __) is a magic method called automatically when an object is about to be destroyed or garbage collected, typically used for cleanup activities like closing database connections or files.

```python
class DatabaseConnection:
    def __init__(self, db_name):
        print(f"Constructor: Connected to database '{db_name}'")
        self.db_name = db_name

    def __del__(self):
        print(f"Destructor: Disconnected from database '{self.db_name}'")


# Object creation (Construction runs)
db = DatabaseConnection("UsersDB")

# Explicitly deleting the object to trigger destructor immediately
del db
```

#### **__new__ vs __init__:** __ new __ is the actual constructor method responsible for creating the object in memory and returning the instance. On the other hand, __ init __ is an initializer that configures the attributes of the object after it has been created. __ new __ takes cls and return a new object, whereas __ init __ takes self and returns None.

```python
class SingletonDemo:
    _instance = None

    # new controls the creation of the object
    def __new__(cls, *args, **kwargs):
        if cls._instance is None:
            print("Creating a new instance using __new__")
            cls._instance = super().__new__(cls)
        return cls._instance

    def __init__(self, value):
        self.value = value

obj1 = SingletonDemo(10)
obj2 = SingletonDemo(20)

print(obj1.value)  # Output: 20 (Kyunki ek hi instance baar-baar use ho raha hai)
print(obj1 is obj2) # Output: True (Singleton Pattern implementation)
```

#### **Class namespace & __ dict __(Internal storage):** In Python, namespaces are mappings from names to objects, implemented internally as dictionaries. Every class and object has a __ dict __ attribute that stores all its writable attributes and their values in a dictionary format.

```python
class Person:
    species = "Homo Sapiens"  # Class Variable

    def __init__(self, name, age):
        self.name = name
        self.age = age

p = Person("Rahul", 25)

# Object ka namespace check karte hain:
print(p.__dict__)
# Output: {'name': 'Rahul', 'age': 25} (Sirf instance variables aate hain)

# Class ka namespace check karte hain:
print(Person.__dict__)
# Output mein 'species' aur baaki class methods/attributes dikhenge.
```

#### **Instance variable:** An Instance variable is a variable that is specific to a particular object (instance). It is defined inside methods (usually *__ init __*) using the self keyword. Every object has its own separate copy of an instance variable, meaning changes made to it in one object do not affect other objects.

```python
class Student:
    def __init__(self, name, age):
        self.name = name  # Instance Variable (Unique to each student)
        self.age = age  # Instance Variable (Unique to each student)

s1 = Student("Amit", 22)
s2 = Student("Rahul", 24)

print(s1.name, s1.age)  # Output: Amit 22
print(s2.name, s2.age)  # Output: Rahul 24
# s1 ka naam badalne se s2 par koi asar nahi padega.
```

#### **Class variable:** A class variable is a variable that is shared among all instances (objects) of a class. It is defined directly inside the class body but outside any methods. If the value of a class variable is changed through the class, it reflects across all objects.

```python
class Student:
    school_name = "Delhi Public School"  # Class Variable (Shared across all)

    def __init__(self, name):
        self.name = name  # Instance variable

s1 = Student("Amit")
s2 = Student("Rahul")

print(s1.school_name)  # Output: Delhi Public School
print(s2.school_name)  # Output: Delhi Public School

# Agar hum class variable ko class ke naam se badlein, toh sabhi ke liye badal jayega:
Student.school_name = "St. Xavier School"
print(s1.school_name)  # Output: St. Xavier School
print(s2.school_name)  # Output: St. Xavier School
```

#### **Class method:**  A class method is bound to the class rather than its object. It takes ***cls*** as its first parameter instead of ***self***, allowing it to access or modify class-level variables. It is created using the ***@classmethod*** decorator.

```python
class Bank:
    bank_interest_rate = 5.5  # class variable

    def __init__(self, account_holder):
        self.account_holder = account_holder

    @classmethod
    def update_interest_rate(cls, new_rate):
        # Class variable ko modify kar rahe hain using 'cls'
        cls.bank_interest_rate = new_rate

print(Bank.bank_interest_rate)  # Output: 5.5

# Class method call kiya
Bank.update_interest_rate(7.0)
print(Bank.bank_interest_rate)  # Output: 7.0(Updated globally)
```

#### **Static method:** A static method is a utility function that belongs to a class but doesn't require access to entire instance (***self***) or (***class***) data. It doesn't take ***self*** or ***cls*** as an   automatic first argument and is defined using the ***@staticmethod*** decorator.

```python
class MathUtils:

    @staticmethod
    def add_numbers(a, b):
        # Na 'self' hai, na 'cls'. Bas ek normal calculation function hai.
        return a + b

# Object banane ki zaroorat nahi padi, direct class se call kar liya!
result = MathUtils.add_numbers(10, 20)
print(result)  # Output: 30
```

```python
class Employee:
    company_name = "TechCorp"  # Class variable

    def __init__(self, name):
        self.name = name  # Instance variable

    @classmethod
    def change_company(cls, new_name):
        cls.company_name = new_name
    
    @staticmethod
    def is_working_day(day):
        return day.lower() not in ["saturday", "sunday"]
```

```python
class Bank:
    bank_name = "Sate Bank Of India"  # Class Variable (shared)

    def __init__(self, owner, balance):
        self.owner = owner  # Instance Variable
        self.balance = balance

    # Class method: Changes Class-level data
    @classmethod
    def update_bank_name(cls, new_name):
        cls.bank_name = new_name

    # Static method: Utility function, no self/cls needed
    @staticmethod
    def validate_age(age):
        return age >= 18

# Usage
print(Bank.bank_name)  # State Bank Of India
Bank.update_bank_name("Python National Bank")
print(Bank.bank_name)  # Python National Bank

print(Bank.validate_age(20))  # True
```
