### Metaclass Kya Hoti Hai? (What is a Metaclass?)
Python mein ek mashhoor rule hai: "Everything is an object". Iska matlab yeh hai ki integers, strings, functions, aur yahan tak ki classes bhi objects hi hain.
* **Normal Object:** Ek normal object kisi Class ka instance hota hai (jaise obj = MyClass()).
* **Class Object:** Ek class object khud kisi Metaclass ka instance hota hai (jaise MyClass kisi metaclass ka instance hai).
###### Python mein by default sabhi classes ki metaclass type hoti hai.

### Built-in type ka Double Role
Hum type() function ko do alag-alag tariko se use karte hain:
1. Type Inspection: Yeh pata lagane ke liye ki koi object kis class ka hai (jaise type(5) output dega <class 'int'>).
2. Dynamic Class Creator: type ko dynamically class banane ke liye bhi use kiya ja sakta hai bina class keyword ke!
### Example (Dynamic Class Creation via type):

```python
Dog = type('Dog', (object, ), {'bark': lambda self: print('Bao Bao')})
mydog = Dog()
mydog.bark()
```

### Class Creation Lifecycle aur __ new __ ka Role
Jab aap Python mein class MyClass: likhte hain, toh Python us class ko turant create nahi kar deta. Iske peeche ek poora lifecycle hota hai:
1. **Parsing & Compilation:** Python code read aur compile hota hai.
2. **Namespace Dictionary Creation:** Class ke andar ke methods aur attributes ki ek dictionary (dict) banti hai.
3. **Metaclass Invocation (__ new __ & __ init __):** Metaclass ka __ new __ method call hota hai jo class object ko memory mein create karta hai, aur phir __ init __ usko initialize karta hai.
###### Aap apni khud ki custom metaclass bana kar is lifecycle ko intercept (beech mein rok kar modify) kar sakte hain!
#### Example (Custom Metaclass using __ new __):

```python
class Meta(type):
    def __new__(cls, name, bases, dct):
        print(f"Class create ho rahi hai: {name}")
        # Yahan hum class create hone se pehle attributes ko modify kar sakte hain
        dct['custom_attribute'] = 'Yeh metaclass ne add kiya hai'
        return super().__new__(cls, name, bases, dct)

# Jab bhi aap yeh class define karenge, metaclass ka __new__ chal padega
class MyClass(metaclass=Meta):
    pass

print(MyClass.custom_attribute)
```

#### Hinglish Explanation: Jab aapne class MyClass(metaclass=Meta) likha, toh Python ne standard class banane ki jagah hamari custom Meta class ke __ new __ method ko call kiya. Isse humne class ke banne se pehle hi usme ek naya attribute (custom_attribute) inject kar diya! Yeh frameworks (jaise Django ORM) mein database fields ko map karne ke liye bohot use hota hai.
