# Kadane's Algorithm

```python
def max_subarray_sum(arr):
    max_sum = float('-inf')
    current_sum = 0

    for num in arr:
        current_sum += num
        if current_sum > max_sum:
            max_sum = current_sum
        if current_sum < 0:
            current_sum = 0

    return max_sum
```

```python
mix_numbers = [3, -2, 5, -1, 6, -3, 2]
result = max_subarray_sum(mix_numbers)

negative_numbers = [-3, -2, -5, -1, -6, -3, -2]
result_negative = max_subarray_sum(negative_numbers)
print("Maximum subarray sum for mixed numbers:", result)
print("Maximum subarray sum for negative numbers:", result_negative)
```

# Kadane's Algorithm using max()

```python
def kadane(arr):
    current_sum = arr[0]
    max_sum = arr[0]
    for num in arr[1:]:
        current_sum = max(num, current_sum + num)
        max_sum = max(max_sum, current_sum)
    return max_sum
```

```python
mix_numbers = [3, -2, 5, -1, 6, -3, 2]
result = kadane(mix_numbers)

negative_numbers = [-3, -2, -5, -1, -6, -3, -2]
result_negative = kadane(negative_numbers)
print("Maximum subarray sum for mixed numbers using Kadane's algorithm:", result)
print("Maximum subarray sum for negative numbers using Kadane's algorithm:", result_negative)
```

# Kadane's Algorithm without max()

```python
def kadane(arr):
    current_sum = 0
    max_sum = arr[0]
    for num in arr[1:]:
        if current_sum + num > num:
            current_sum += num
        else:
            current_sum = num
        if current_sum > max_sum:
            max_sum = current_sum
    return max_sum
```

```python
mix_numbers = [3, -2, 5, -1, 6, -3, 2]
result = kadane(mix_numbers)
negative_numbers = [-3, -2, -5, -1, -6, -3, -2]
result_negative = kadane(negative_numbers)
print("Maximum subarray sum for mixed numbers using Kadane's algorithm:", result)
print("Maximum subarray sum for negative numbers using Kadane's algorithm:", result_negative)
```
